**lambda function**

- lambda function is a function in a single line

- many arguments but only one expression

```python
lambda  <arguments>:<expression>
```

```python
def add(a):
    print(a+10)

add(100)
```

```python
add1=lambda a : a+10
add1(20)
```

```python
cube=lambda x : x**3
cube(10)
```

```python
avg=lambda a,b,c : round(a+b+c)/3
avg(10,20,30)
```

```python
avg1=lambda a,b,c=50 : round((a+b+c)/3,2)
avg1(20,30)
```

**if -else**

```python
lamda <arguments> : <if_out> (if_con>) else <else_out> 
```

```python
even_odd=lambda a : 'even' if a%2==0 else 'odd'
even_odd(10)
```

```python
 #in lambda function only if condition is not valid
```

```python
max=lambda a,b : a if a>b else b
max(2,5)
```
