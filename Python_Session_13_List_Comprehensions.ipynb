- List Comprehensions are used to write in single line operation

```python
l=[]
for i in range(10):
    l.append(i)

l
```

```python
# l=[<output> <forloop>]
l=[i for i in range(10)]
l
```

```python
l=['hyd','mumbai','pune']
l=[i.upper() for i in  l ]
l
```

**for -if condition**

```python
l=[20,23,27,44,46]
l=[i for i in l if i%2==0]
l
```

```python
l=['hyd','mumbai','pune']
l=[i for i in l if len(i)>3]
l
```

**for if else**

```python
[ <if out>  <if cond> <else out> <else cond> <for loop>]
```

```python
l=[20,23,27,44,46]
l=[f'{i}_even' if i%2==0 else f'{i}_odd' for i in l]
l
```

```python
l=[44,67,78,97]
out=[]
for i in l:
    if i>75:
        out.append('A')
    elif i>65:
        out.append('B')
    elif i>97:
        out.append('C')
    else:
        out.append('D')
out
```
