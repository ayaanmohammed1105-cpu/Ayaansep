*args and **kwargs

- args means arguments we can pass multiple arguments at a time 

  - output come in a list

- kwargs means keyword arguments, we can pass multiple arguments at a time 

  - but we need to mention variable(key)  name 

  - output come in  a dictionary

```python
def f1(*args):
    print(args)
```

```python
f1(10,20)
```

```python
def f2(**kwargs):
    print(kwargs)
```

```python
f2(a=10,b=20)
```

```python
def f1(**kwargs):
    print(kwargs)

f1(NAME='NARESHIT',AGE='21',CITY='HYDERABAD')
```
